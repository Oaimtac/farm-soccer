# 1. 以蒐集來的數據，建立PAT與BP的對應資料庫與一元一次迴歸方程式

> 模擬算式並非難事，只需要紀錄PAT時搭配血壓計量測，將兩方資料進行迴歸計算即可

> 由於同學們可能身邊沒有血壓計，以範例的資料舉例說明

> 血壓估算一元一次方程式：BP = A * PAT + B

```python
#以下範例的資料，以血壓計建立的十筆資料，舉例量測血壓110mm/Hg時的資料放入BP第一個陣列點，對應求出的PAT為241.38毫秒，放入PAT第一個陣列點
#若想建立自己的模型，直接修改這邊的資料即可

BP = np.array([110, 127, 115, 120, 131, 125, 117, 121, 120, 126])    #單位：毫米汞柱

PAT = np.array([241.38, 228.85, 233.43, 223.22, 214.45, 225.2, 218.68, 219.19, 255, 236.93]) #單位：毫秒
```

```python
#將BP和PAT進行迴歸計算！
lm = LinearRegression()  #呼叫一個迴歸係數計算機
lm.fit(PAT, BP)
```

```python
#將BP和PAT進行迴歸計算！
lm = LinearRegression()  #呼叫一個迴歸係數計算機
lm.fit(PAT, BP)
```

```python
#將一維陣列轉換成二維矩陣的方法：array.reshape(column, row)
PAT_matrix = PAT.reshape(len(PAT), 1) #建立一個n*1的PAT資料矩陣
BP_matrix = BP.reshape(len(BP), 1) #建立一個n*1的PAT資料矩陣
print("PAT資料矩陣:", PAT_matrix)
print("BP資料矩陣:", BP_matrix)
```

```python
#將BP和PAT進行迴歸計算！
lm = LinearRegression()  #呼叫一個迴歸係數計算機
lm.fit(PAT_matrix, BP_matrix)

print("一元一次方程式的係數A為:", lm.coef_)
print("一元一次方程式的截距B為:", lm.intercept_)

print('迴歸計算出一元一次方程式為 BP=',lm.coef_[0,0],'* PAT +',lm.intercept_[0]) #迴歸係數計算機的資料結果型態較特殊，請留意取法
```
