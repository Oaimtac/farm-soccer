#1-3條件(Condition: if/else)、布林值(Boolean: true/false)及其他運算子(Operators)


常見運算子：

1. 反邏輯(not): not
2. 並且(and): and、&
3. 或者(or): or、|
4. 是否相等: ==
5. 是否不相等：!=


```python
條件 = True # or Fasle
if(條件):
  print("真")
else:
  print("假")
```

```python
條件 = True # or Fasle
if(not 條件):
  print("真")
else:
  print("假")
```

```python
條件一 = True
條件二 = False
if(條件一 and 條件二):
  print("真")
else:
  print("假")
```

```python
條件一 = True
條件二 = False
if(條件一 or 條件二):
  print("真")
else:
  print("假")
```

```python
x = 1
y = 1
if(x == y):
  print("真")
else:
  print("假")
```

```python
x = 1
y = 1
if(x != y):
  print("真")
else:
  print("假")
```

```python
x = 1
y = 2
if(x == y):
  print("真")
else:
  print("假")
```

```python
x = 1
y = 2
if(x != y):
  print("真")
else:
  print("假")
```

```python
sign = input("sign is:")
if(sign == "+"):
  print("加號")
elif(sign == "-"):
  print("減號")
elif(sign == "*"):
  print("乘號")
elif(sign == "/"):
  print("除號")
else:
  print("非四則運算符號")
```

```python
myString = "How long"
print("此段字串的長度為:",len(myString))
print(myString[0])
print(myString[1])
print(myString[2])
print(myString[3])
print(myString[4])
print(myString[5])
print(myString[6])
print(myString[7])
```

```python
myString = "How long"
myString = myString + " is the String"
print(myString)
```

```python
myString = "很重要所以講三次\n" * 3
print(myString)
```

```python
myString = "abcdefghijklmnopqrtuvwxyz"
print('w' in myString)
print('s' in myString)

if('w' in myString):
  print("There's a [w] in myString")
else:
  print("There's no [w] in myString")

if('s' in myString):
  print("There's a [s] in myString")
else:
  print("There's no [s] in myString")
```
