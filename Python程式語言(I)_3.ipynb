```python
file = open('Python程式語言(I)_3_石板密文.txt', mode='r') #以file來執行「開啟石板密文檔案」
lines = file.readlines()                  #以lines來讀取file中的「所有行數文字」
print(lines)   #印出lines來查看「所有行數文字」
print("\n")   #換行
print("密文第一行：",lines[0]) #印出lines的第一行文字
print("密文第二行：",lines[1]) #印出lines的第二行文字
print("密文第三行：",lines[2]) #印出lines的第三行文字
print("密文第四行：",lines[3]) #印出lines的第四行文字
file.close()  #完成讀取，file「關閉石板密文檔案」
```

```python
#發現寶藏室前倒地的人，手中握有片段密文解密，上面寫著...
#將密文第一行做為解密關鍵文，將第二行的數字對應，似乎能得到一些提示！

secret_number = lines[0]       #將密文第一行做為解密關鍵文記錄下來
print("密文第一行化成解密文字：",secret_number)  #印出解密關鍵第一行密文

locate1 = 0          #lines的第二行文字中呈現的第一個數字
locate2 = 17         #lines的第二行文字中呈現的第二個數字
locate3 = 18         #lines的第二行文字中呈現的第三個數字


print("第二行文字0，搭配到解密關鍵文，得出：",secret_number[locate1])
print("第二行文字17，搭配到解密關鍵文，得出：",secret_number[locate2])
print("第二行文字18，搭配到解密關鍵文，得出：",secret_number[locate3])
print("將三個密文結合起來發現：",secret_number[locate1], secret_number[locate2], secret_number[locate3])
print("難道說...解開密碼的方法...，就是將第三行、第四行文字搭配解密關鍵文印出來！？")
```
