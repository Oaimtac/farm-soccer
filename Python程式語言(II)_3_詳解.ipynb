#2-5 產生等差數列的好幫手、range的進化版：Numpy linspace



> range數列：range(起始, 終點, 步長)

> Numpy linspace數列：np.linspace(起始, 終點, 個數)



```python
import numpy as np
x1 = range(0, 10, 1)
x2 = np.linspace(0, 9, 10)
print("x1是range產生的數列，型別是:", type(x1), "印出為：",x1)
print("x2是np.linspace產生的數列，型別是:", type(x2), "印出為：",x2)
print("\n")

print("x1[0]是range產生的數值，型別是:", type(x1[0]), "印出為：",x1[0])
print("x2[0]是np.linspace產生的數值，型別是:", type(x2[0]), "印出為：",x2[0])
```

```python
x1 = range(0, 10, 1)
x2 = np.linspace(0, 9, 10)
print("x1用for迴圈逐一印出為：")
for i in x1:
  print(i, end=" ")
print("\n")

print("x2用for迴圈逐一印出為：")
for i in x2:
  print(i, end=" ")
```

```python
#np.linspace在浮點數也能產生等差數列

z1 = np.linspace(0, 1, 10)
print("從0到1，設定等差數列個數10個:")
print("z1:", z1)
```

```python
#np.linspace在浮點數也能產生等差數列
z2 = np.linspace(0, 1, 11)
print("從0到1，設定等差數列個數11個:")
print("z1:", z2)
```

```python
#Numpy的陣列(array)可以非常便利的運算與擷取，賦予np.linspace強大的功能
#功能1: 數列擷取

y0 = np.linspace(0, 9, 10)
y1 = y0[0:5]

print("y0:",y0)
print("y1:",y1)
```

```python
#Numpy的陣列(array)可以非常便利的運算與擷取，賦予np.linspace強大的功能
#功能2: 數列運算-乘法
y0 = np.linspace(0, 9, 10)
y2 = y0*2

print("y0:",y0)
print("y2:",y2)
```

```python
#Numpy的陣列(array)可以非常便利的運算與擷取，賦予np.linspace強大的功能
#功能3: 數列運算-加法
y0 = np.linspace(0, 9, 10)
y3 = y0 + 1
print("y0:",y0)
print("y3:",y3)
```

```python
#Numpy的陣列(array)可以非常便利的運算與擷取，賦予np.linspace強大的功能
#功能4: 數列全選
y0 = np.linspace(0, 9, 10)
y4 = y0[:]
print("y0:",y0)
print("y4:",y4)
```

```python
pi = np.pi            #利用numpy幫我們建立的pi來建立一個pi數值 (pi = 180度)
pi_5 = round(pi, 5)       #利用round函式幫我們取小數點後5位
sin_30 = round(np.sin(pi/6), 3) #利用round函式幫我們取小數點後3位，避免不必要誤差
sin_90 = round(np.sin(pi/2), 3)
sin_150 = round(np.sin(pi/1.2), 3)
sin_180 = round(np.sin(pi), 3)

print("pi:",pi)
print("pi取小數點後五位:",pi_5)
print("sin 30度:",sin_30)
print("sin 90度:",sin_90)
print("sin 150度:",sin_150)
print("sin 180度:",sin_180)
```



> 搭配圖形功能，用程式畫出一個好看的sin波！

1.   準備畫布與座標
2.   以x軸為時間單位, y軸為sin波數值
3.   搭配numpy linspace將sin的波形完整描繪出來


```python
import matplotlib.pylab as plt     #從matplotlib函式庫中，引入pylab函式，命名為plt
dots = 10                #嘗試調整dots大小，觀察對sin波的影響
time = np.linspace(0, 1, dots)     #宣告一個從0到1之間的等差數列
pi_time = 2*np.pi*time          #使用上方數列，乘上2pi，建立對應的pi陣列
y_sin = np.sin(pi_time)          #使用上方宣告的pi陣列，建立對應的sin函數陣列

fig, ax1 = plt.subplots(1 ,1)  #透過plt建立畫布fig與座標軸ax1
ax1.plot(time, y_sin)        #將time, y數值代入座標軸ax1
ax1.set_xlabel('Time [seconds]')  #設定x軸座標為時間
plt.show()              #秀出畫布
```

#2-5(exercise2)請創造一個2秒、頻率為5赫茲(Hz)且波形完整的sin波

```python
#範例解答
dots = 1000             #足夠多的dots，才能使sin波波形完整
hz = 5               #宣告一個赫茲(hz)參數
time = np.linspace(0, 2, dots)  # 宣告0到2之間的等差數列，使最終秒數為2
pi_time = 2*np.pi*time*hz      ##使用上方數列，乘上2pi，建立對應的pi陣列，進一步乘上hz，使sin函數能循環更快，產生更高頻率的sin波
y_sin = np.sin(pi_time)       #使用上方宣告的pi陣列，建立對應的sin函數陣列

fig, ax1 = plt.subplots(1 ,1)  #透過plt建立畫布fig與座標軸ax1
ax1.plot(time, y_sin)        #將time, y數值代入座標軸ax1
ax1.set_xlabel('Time [seconds]')  #設定x軸座標為時間
plt.show()              #秀出畫布
```
